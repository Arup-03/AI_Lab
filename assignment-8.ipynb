```python
# Role-Biased Expert System for Student Performance Evaluation

def evaluate_student(attendance, marks, assignment, internal):
    """
    Evaluates student performance using IF-THEN rules.
    All inputs are assumed to be percentages (0-100).
    """

    # IF-THEN rules
    if (attendance >= 90 and marks >= 85 and
            assignment >= 85 and internal >= 85):
        result = "Excellent"

    elif (attendance >= 80 and marks >= 70 and
          assignment >= 70 and internal >= 70):
        result = "Good"

    elif (attendance >= 65 and marks >= 50 and
          assignment >= 50 and internal >= 50):
        result = "Average"

    else:
        result = "Poor"

    return result


# Main program
print("==========================================")
print(" Student Performance Expert System")
print("==========================================")

# Taking inputs
attendance = float(input("Enter attendance percentage: "))
marks = float(input("Enter marks percentage: "))
assignment = float(input("Enter assignment performance percentage: "))
internal = float(input("Enter internal assessment percentage: "))

# Validate input
if not (0 <= attendance <= 100 and
        0 <= marks <= 100 and
        0 <= assignment <= 100 and
        0 <= internal <= 100):

    print("\nError: Please enter values between 0 and 100.")

else:
    # Evaluate student
    performance = evaluate_student(
        attendance, marks, assignment, internal
    )

    # Display result
    print("\n----------- Evaluation Result -----------")
    print(f"Attendance          : {attendance}%")
    print(f"Marks               : {marks}%")
    print(f"Assignment          : {assignment}%")
    print(f"Internal Assessment : {internal}%")
    print(f"Performance         : {performance}")
```

Describe a python to implement a rule-based expert system for medical diagnosis.The system should accept symptoms such as fever,cough,hedache,and sore throat and use IF THEN rule to identify a possible disease.

```python

# Rule-Based Expert System for Medical Diagnosis

def diagnose(fever, cough, headache, sore_throat):

    # IF-THEN rules

    # Rule 1: Common Cold
    if cough and sore_throat and not fever:
        disease = "Common Cold"

    # Rule 2: Flu
    elif fever and cough and headache:
        disease = "Influenza (Flu)"

    # Rule 3: Possible Strep Throat
    elif fever and sore_throat and not cough:
        disease = "Possible Strep Throat"

    # Rule 4: Possible COVID-19
    elif fever and cough and sore_throat:
        disease = "Possible COVID-19"

    # Rule 5: Unknown condition
    else:
        disease = "Unable to determine"

    return disease


# Main program
print("======================================")
print(" Medical Diagnosis Expert System")
print("======================================")
print("Answer with yes or no.\n")


def get_answer(question):
    answer = input(question + " (+yes/no): ").lower()

    while answer not in ["yes", "no"]:
        print("Please enter yes or no.")
        answer = input(question + " (yes/no): ").lower()

    return answer == "yes"


# Get symptoms
fever = get_answer("Do you have fever?")
cough = get_answer("Do you have cough?")
headache = get_answer("Do you have headache?")
sore_throat = get_answer("Do you have sore throat?")

 
# Diagnose
result = diagnose(
    fever,
    cough,
    headache,
    sore_throat
)

print("\n========== Result ==========")
print("Possible condition:", result)
print("============================")
print("\nThis result is for educational purposes only.")
print("Consult a qualified healthcare professional for medical advice.")
```
