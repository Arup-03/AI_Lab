write a python program to implement 8 puzzle without heuristic function using BFS algorithm

```python
from collections import deque

start = [
    [1, 2, 3],
    [4, 0, 6],
    [7, 5, 8]
]

goal = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0]
]


def find_blank(state):
    for i in range(3):
        for j in range(3):
            if state[i][j] == 0:
                return i, j


def state_to_tuple(state):
    new_state = []

    for row in state:
        new_state.append(tuple(row))

    return tuple(new_state)


def get_neighbors(state):
    neighbors = []

    
    x, y = find_blank(state)

    
    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dx, dy in moves:
        nx = x + dx
        ny = y + dy

        # Check if new position is inside the puzzle
        if 0 <= nx < 3 and 0 <= ny < 3:

        
            new_state = [row[:] for row in state]

            
            new_state[x][y], new_state[nx][ny] = new_state[nx][ny], new_state[x][y]

            
            neighbors.append(new_state)

    return neighbors


def bfs(start):
    queue = deque()
    queue.append((start, []))

    visited = set()

    while queue:

        current, path = queue.popleft()

        
        if current == goal:
            return path + [current]

        current_tuple = state_to_tuple(current)

        if current_tuple not in visited:

            visited.add(current_tuple)

            for neighbor in get_neighbors(current):
                queue.append((neighbor, path + [current]))

    return None


def print_solution(solution):
    print("Solution Found!\n")

    for step, state in enumerate(solution):
        print("Step", step)

        for row in state:
            print(row)

        print()


solution = bfs(start)

if solution:
    print_solution(solution)
else:
    print("No Solution Exists.")
```
