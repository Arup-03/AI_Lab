Write a python program to implement the minimax algorithm for a two-player game.

```python
import math

def minimax(depth, node_index, is_maximizing, values, max_depth):
    if depth == max_depth:
        return values[node_index]

    if is_maximizing:
        return max(
            minimax(depth+1, node_index*2, False, values, max_depth),
            minimax(depth+1, node_index*2+1, False, values, max_depth)
        )
    else:
        return min(
            minimax(depth+1, node_index*2, True, values, max_depth),
            minimax(depth+1, node_index*2+1, True, values, max_depth)
        )

leaf_values = [3, 5, 2, 9, 12, 5, 23, 23]
tree_depth = int(math.log2(len(leaf_values)))

best_score = minimax(0, 0, True, leaf_values, tree_depth)
print("Best possible score:", best_score)
```

```python
import math

leaf_values = [3, 5, 2, 9, 12, 5, 23, 23]

def minimax(depth, node_index, maximizing_player, values, max_depth):

    if depth == max_depth:
        return values[node_index]

    if maximizing_player:
        best = float('-inf')

        for i in range(2):
            score = minimax(
                depth + 1,
                node_index * 2 + i,
                False,
                values,
                max_depth
            )
            best = max(best, score)

        return best

    else:
        best = float('inf')

        for i in range(2):
            score = minimax(
                depth + 1,
                node_index * 2 + i,
                True,
                values,
                max_depth
            )
            best = min(best, score)

        return best


tree_depth = int(math.log2(len(leaf_values)))

best_score = minimax(0, 0, True, leaf_values, tree_depth)

print("Best possible score:", best_score)
```

write a python program to implement thr Alpha Beta pruning algorithm for a two player game.use the leaf node values 3,5,8,12,5,23,23 . determine the optimal score for the max player and display which branches are pruned during the search

```python
import math

# Leaf node values
values = [3, 5, 8, 12, 5, 23, 23]

def alpha_beta(depth, node_index, maximizing_player, alpha, beta, max_depth):

    # If leaf node is reached
    if depth == max_depth:
        return values[node_index]

    if maximizing_player:
        best = float('-inf')

        for i in range(2):
            score = alpha_beta(
                depth + 1,
                node_index * 2 + i,
                False,
                alpha,
                beta,
                max_depth
            )

            best = max(best, score)
            alpha = max(alpha, best)

            # Alpha-Beta pruning
            if beta <= alpha:
                print("Pruned branch at node:", node_index * 2 + i)
                break

        return best

    else:
        best = float('inf')

        for i in range(2):
            score = alpha_beta(
                depth + 1,
                node_index * 2 + i,
                True,
                alpha,
                beta,
                max_depth
            )

            best = min(best, score)
            beta = min(beta, best)

            # Alpha-Beta pruning
            if beta <= alpha:
                print("Pruned branch at node:", node_index * 2 + i)
                break

        return best


# Calculate tree depth
tree_depth = math.ceil(math.log2(len(values)))

# Start Alpha-Beta pruning
best_score = alpha_beta(
    0,
    0,
    True,
    float('-inf'),
    float('inf'),
    tree_depth
)

print("\nOptimal score for MAX player:", best_score)
```
